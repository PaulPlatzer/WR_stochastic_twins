<h1><center>Illustration: sources of changes in estimated local dimension</center></h1>

This notebook was written by Paul Platzer. It accompanies the following scientific publication:<br>
"Disentangling Density and Geometry in Weather Regime Dimensions using Stochastic Twins"<br>
By Paul Platzer, Bertrand Chapron, and Gabriele Messori (2025).<br>
<b> https://doi.org/10.21203/rs.3.rs-5880040/v1 </b>

It is part of seven notebooks that allow to reproduce the figures of the article. This notebook provides illustrational examples of how the estimated dimension may be influenced either by changes in the local geometry or by changes in the local density inside the ball that is used to find analogues which will be used to estimate the local dimension.

What this notebook does:

<ol>
    <li> It generates data from 4 different distributions in a squared box of the plane. For each distribution, 100 datasets are generated, and each dataset is made of 4000 samples.</li>
    <li> It computes the local dimension at the origin point (0,0) for all possible numbers of analogues (within the ball of radius 1) and for all generated datasets. This allows to provide confidence intervals (as quantiles) for the estimated local dimension and for each number of analogues and each distribution.</li>
    <li> It plots the results.</li>
</ol>

Note that, for this code to work on your machine, you should:

<ol>
    <li> Install the libraries listed in the first two cells. </li>
    <li> Use python 3 : this code was tested using python 3.10.12 (main, Nov  6 2024, 20:22:13) [GCC 11.4.0] </li>
    <li> Have enough memory and computing resources. However, this code is pretty light. If necessary, you could set the number of datasets "$M$" as an argument inside the function <code>compare_plots()</code>. </li>
</ol>
